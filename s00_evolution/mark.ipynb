openharness/  
├── engine/        # Agent主循环引擎  
├── tools/         # 43个内置工具实现  
├── skills/        # Markdown按需加载技能  
├── plugins/       # 插件扩展（兼容Claude插件）  
├── permissions/   # 权限安全管控  
├── hooks/         # 工具执行生命周期钩子  
├── commands/      # 54条内置斜杠命令 /commit /plan /skills  
├── mcp/           # Model Context Protocol客户端   
├── memory/        # 持久记忆、上下文压缩  
├── coordinator/   # 多Agent协同调度  

我们可以简要的把claude code的架构分为十大部分：  
- 1. Agent loop：Agent主循环引擎，负责处理输入、执行工具、与用户交互等。
- 2. tools：内置的工具实现，如文件操作、代码生成、代码分析等。
- 3. memory：持久记忆、上下文压缩等。
- 4. coordinator：多Agent协同调度模块，用于协调多个Agent之间的合作。
- 5. 权限安全管控：权限安全管控模块，负责管理用户权限、访问控制等。  
  
      上面的模块都是基础模块，是能够让claude code的基本功能正常运行；  
      下面的模块都是扩展模块，能够提供额外的功能和灵活性。

- 6. skills：Markdown按需加载技能，用于处理复杂的任务。
- 7. plugins：插件扩展模块，兼容Claude插件。
- 8. hooks：工具执行生命周期钩子模块，用于在工具执行前后执行自定义操作。
- 9. commands：54条内置斜杠命令 /commit /plan /skills等。
- 10. mcp：Model Context Protocol客户端，用于与模型进行交互。
